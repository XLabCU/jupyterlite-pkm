## Creating a DataFrame

A DataFrame is a 2D data structure, like a spreadsheet.

```python
# Create a simple DataFrame
data = {
    'Name': ['Alice', 'Bob', 'Charlie', 'Diana'],
    'Age': [25, 30, 35, 28],
    'City': ['New York', 'Paris', 'London', 'Tokyo'],
    'Score': [85, 90, 78, 92]
}

df = pd.DataFrame(data)
df
```

## Inspecting Data

```python
# View first few rows
print("First 2 rows:")
print(df.head(2))

print("\nDataFrame info:")
print(df.info())
```

```python
# Statistical summary
df.describe()
```

## Filtering Data

```python
# Filter rows where Score > 85
high_scorers = df[df['Score'] > 85]
print("High scorers (Score > 85):")
high_scorers
```
