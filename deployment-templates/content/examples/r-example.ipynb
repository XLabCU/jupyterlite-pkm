## Working with Data Frames

```r
# Create a data frame
df <- data.frame(
  name = c("Alice", "Bob", "Charlie"),
  age = c(25, 30, 35),
  score = c(85, 90, 78)
)

print(df)
```

## Basic Plotting

```r
# Simple plot
x <- 1:10
y <- x^2

plot(x, y, 
     type = "b",
     main = "Quadratic Function",
     xlab = "X",
     ylab = "Y = X^2",
     col = "blue",
     pch = 19)
```

## Using ggplot2

First, install ggplot2 if not already available:

```r
# Install and load ggplot2 (may take a moment)
webr::install("ggplot2")
library(ggplot2)
```

```r
# Create a ggplot
df <- data.frame(
  category = c("A", "B", "C", "D"),
  value = c(3, 7, 2, 5)
)

ggplot(df, aes(x = category, y = value, fill = category)) +
  geom_bar(stat = "identity") +
  theme_minimal() +
  labs(title = "Bar Chart Example")
```
